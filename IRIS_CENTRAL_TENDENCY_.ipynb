```python
import pandas as pd
df = pd.read_csv("iris.csv")
print("\nFirst 5 rows:")
print(df.head())
print("\nColumn Names:")
print(df.columns)
numeric_columns = df.select_dtypes(include='number').columns
print(numeric_columns)

# Step 7: Calculate Mean
print("\n----- MEAN -----")
print(df[numeric_columns].mean())

# Step 8: Calculate Median
print("\n----- MEDIAN -----")
print(df[numeric_columns].median())

# Step 9: Calculate Mode
print("\n----- MODE -----")
print(df[numeric_columns].mode())

# Step 10: Calculate Minimum
print("\n----- MINIMUM -----")
print(df[numeric_columns].min())

# Step 11: Calculate Maximum
print("\n----- MAXIMUM -----")
print(df[numeric_columns].max())
```
