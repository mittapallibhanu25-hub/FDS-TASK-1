```python
# ============================================
# DATA WRANGLING USING PANDAS
# MERGE, CONCAT AND JOIN
# ============================================

import pandas as pd

# ============================================
# 1. CREATE EMPLOYEE DATAFRAME 1
# ============================================

emp1 = pd.DataFrame({
    "Emp_ID": [101, 102, 103, 104, 105],
    "Name": ["Mithali", "Samiksha", "Jyothish", "Tanish", "Suhani"],
    "Dept": ["IT", "HR", "Finance", "IT", "Marketing"]
})

print("===== EMPLOYEE DATAFRAME 1 =====")
print(emp1)


# ============================================
# 2. CREATE EMPLOYEE DATAFRAME 2
# ============================================

emp2 = pd.DataFrame({
    "Emp_ID": [106, 107, 108, 109, 110],
    "Name": ["Praneeth", "Ashish", "Lohit", "Tharun", "Ganesh"],
    "Dept": ["HR", "IT", "Finance", "Marketing", "IT"]
})

print("\n===== EMPLOYEE DATAFRAME 2 =====")
print(emp2)


# ============================================
# 3. CREATE SALARY DATAFRAME
# ============================================

salary = pd.DataFrame({
    "Emp_ID": [101, 102, 103, 104, 105,
               106, 107, 108, 109, 110],
    "Salary": [45000, 40000, 55000, 50000, 42000,
               48000, 46000, 58000, 44000, 52000]
})

print("\n===== SALARY DATAFRAME =====")
print(salary)


# ============================================
# 4. CONCATENATION
# Combine emp1 and emp2 row-wise
# ============================================

emp = pd.concat([emp1, emp2], ignore_index=True)

print("\n===== CONCATENATION =====")
print(emp)


# ============================================
# 5. CONCATENATION COLUMN-WISE
# ============================================

df1 = pd.DataFrame({
    "Name": ["Tanish", "Praneeth", "Ashish"],
    "Age": [20, 21, 20]
})

df2 = pd.DataFrame({
    "Marks": [85, 78, 92],
    "Grade": ["A", "B", "A"]
})

concat_columns = pd.concat([df1, df2], axis=1)

print("\n===== COLUMN-WISE CONCATENATION =====")
print(concat_columns)


# ============================================
# 6. INNER MERGE
# Only matching Emp_ID values
# ============================================

inner_merge = pd.merge(
    emp,
    salary,
    on="Emp_ID",
    how="inner"
)

print("\n===== INNER MERGE =====")
print(inner_merge)


# ============================================
# 7. LEFT MERGE
# All records from left DataFrame
# ============================================

left_merge = pd.merge(
    emp,
    salary,
    on="Emp_ID",
    how="left"
)

print("\n===== LEFT MERGE =====")
print(left_merge)


# ============================================
# 8. RIGHT MERGE
# All records from right DataFrame
# ============================================

right_merge = pd.merge(
    emp,
    salary,
    on="Emp_ID",
    how="right"
)

print("\n===== RIGHT MERGE =====")
print(right_merge)


# ============================================
# 9. OUTER MERGE
# All records from both DataFrames
# ============================================

outer_merge = pd.merge(
    emp,
    salary,
    on="Emp_ID",
    how="outer"
)

print("\n===== OUTER MERGE =====")
print(outer_merge)


# ============================================
# 10. MERGE USING LEFT_ON AND RIGHT_ON
# ============================================

employee = pd.DataFrame({
    "Employee_ID": [101, 102, 103],
    "Name": ["Tanish", "Praneeth", "Ashish"]
})

employee_salary = pd.DataFrame({
    "ID": [101, 102, 103],
    "Salary": [50000, 48000, 46000]
})

merge_different_columns = pd.merge(
    employee,
    employee_salary,
    left_on="Employee_ID",
    right_on="ID"
)

print("\n===== MERGE USING DIFFERENT COLUMN NAMES =====")
print(merge_different_columns)


# ============================================
# 11. JOIN
# Join DataFrames using INDEX
# ============================================

emp_details = pd.DataFrame({
    "Name": ["Tanish", "Praneeth", "Ashish"],
    "Dept": ["IT", "HR", "Finance"]
}, index=[101, 102, 103])


emp_salary = pd.DataFrame({
    "Salary": [50000, 48000, 46000]
}, index=[101, 102, 103])


join_result = emp_details.join(emp_salary)

print("\n===== JOIN =====")
print(join_result)


# ============================================
# 12. LEFT JOIN
# ============================================

left_join = emp_details.join(
    emp_salary,
    how="left"
)

print("\n===== LEFT JOIN =====")
print(left_join)


# ============================================
# 13. RIGHT JOIN
# ============================================

right_join = emp_details.join(
    emp_salary,
    how="right"
)

print("\n===== RIGHT JOIN =====")
print(right_join)


# ============================================
# 14. OUTER JOIN
# ============================================

outer_join = emp_details.join(
    emp_salary,
    how="outer"
)

print("\n===== OUTER JOIN =====")
print(outer_join)


# ============================================
# 15. FINAL EMPLOYEE DATA WITH SALARY
# ============================================

final_df = pd.merge(
    emp,
    salary,
    on="Emp_ID",
    how="inner"
)

print("\n===== FINAL EMPLOYEE DATA =====")
print(final_df)
```
